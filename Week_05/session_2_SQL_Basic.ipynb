# Day 1: SQL Basics & Connecting with SQLite

Welcome to your SQL Sprint! Today, we'll learn the basics of SQL (Structured Query Language) and how to use it right inside this notebook with Python's built-in `sqlite3` library.

### 1. Setup and Imports

```python
import sqlite3
import pandas as pd

print(f"Using SQLite version: {sqlite3.sqlite_version}")
```

### 2. Core Concept: What is SQLite?

- It's a database engine that is server-less, self-contained, and requires no configuration.
- The entire database (definitions, tables, data) is stored in a single file, which is perfect for learning and small projects.
- Python has a built-in library, `sqlite3`, to work with it.

### 3. Task 1: Connect & Create a Database

We'll create a connection (`conn`) to a database file. If the file doesn't exist, SQLite will create it. We also create a `cursor` (`c`), which is used to execute SQL commands.

```python
conn = sqlite3.connect('db/company.db')
c = conn.cursor()
```

### 4. Task 2: CREATE Table

Let's create our first table. SQL commands are written as strings. We'll make an `employees` table. 

Common SQL Data Types in SQLite:
- `TEXT`: For strings (e.g., 'John Doe')
- `INTEGER`: For whole numbers (e.g., 42)
- `REAL`: For floating-point numbers (e.g., 65000.50)
- `INTEGER PRIMARY KEY`: A special column that must be unique and is often used as an ID.

```python
c.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY, 
    name TEXT NOT NULL, 
    department TEXT, 
    salary REAL
)
""")

print("Table 'employees' created (or already exists).")
```

### 5. Task 3: INSERT Data

Now let's add some rows (records) to our table. After we make a change (like `INSERT`, `UPDATE`, `DELETE`), we must call `conn.commit()` to save the change to the file.

```python
c.execute("INSERT INTO employees (name, department, salary) VALUES ('Alice Smith', 'Engineering', 90000)")

conn.commit()

print("Added Alice Smith.")
```

**Task (Challenge):** Use `executemany` to insert multiple records at once. This is much more efficient.

```python
new_employees = [
    ('Bob Johnson', 'Sales', 65000),
    ('Charlie Brown', 'Engineering', 110000),
    ('David Lee', 'Sales', 70000),
    ('Eva Grant', 'HR', 55000)
]

# The '?' are placeholders
c.executemany("INSERT INTO employees (name, department, salary) VALUES (?, ?, ?)", new_employees)

conn.commit()
print(f"Added {len(new_employees)} new employees.")
```

### 6. Task 4: SELECT Data (Querying)

This is the most common SQL command. It's used to *read* data. `SELECT *` means "select all columns".

```python
c.execute("SELECT * FROM employees")

all_data = c.fetchall()

for row in all_data:
    print(row)
```

### 7. Task 5: The WHERE Clause (Filtering)

`WHERE` is used to filter records. It's the SQL equivalent of a `pandas` boolean mask (e.g., `df[df['department'] == 'Sales']`).

```python
print("--- Engineering Team ---")
c.execute("SELECT name, salary FROM employees WHERE department = 'Engineering'")
engineering_team = c.fetchall()
for row in engineering_team:
    print(row)

print("\n--- High Earners (Salary > 80k) ---")
c.execute("SELECT * FROM employees WHERE salary > 80000")
high_earners = c.fetchall()
for row in high_earners:
    print(row)
```

### 8. Cleanup: Close Connection

It's good practice to close the connection when you're done.

```python
conn.close()
print("Connection closed.")
```
